## The solution are **Recurent** **Neural** **Networks**, at least for today.
### What are they? It is a special type of Neural Networks made to process sequences of data, with a time step ranging from 1 to N. They are called **Recurrent** because the output of each element is dependent on the previous one.

### Architecture :     
<img src=https://miro.medium.com/v2/resize:fit:1186/format:webp/1*JOkrQoJ3J3-451GzRcayRg.png widht= 500 height=500>

## The input is processed from right to the left side. Lets see an example, if the sequence is a sentence of 3 words the RNN would have 3 layers for each element. Strange to think for it as layers, so think as elements for each word


## 1.   Input : x(t)  t is a timestep
## 2.   Hidden state: h(t) represents a hidden state at time t and acts as "memory". The Calculation inside each hidden state is h(t) = f(**U** * x(t) + **W** * h(t-1)+ Bias) The F is an activation function which should be non-linear transformation such as tanh or Relu.
## 3.   Weights: Weight matrix named **U** which is parametrizing the input connection x(t). Weight matrix **W** which is connecting the previous state (t). Finally the weight V which parametrizes the output of the hidden state.
## 4.   Output: o(t) illustrates the output of the network. In this figure you can visualize the output which is also subject to non linearity because many tasks must have another layers after the RNN.
